```python
import pandas as pd

data = {
    'state': ['Punjab', 'UP', 'Maharashtra', 'Tamil Nadu', 'Bihar', 'Gujarat', 'West Bengal'],
    'crop': ['Wheat', 'Rice', 'Sugarcane', 'Rice', 'Wheat', 'Cotton', 'Rice'],
    'yield_kg_per_ha': [4500, 2800, 7200, 3100, 2200, 3500, 3000],
    'rainfall_mm': [650, 900, 1100, 1400, 1050, 700, 1500]
}

df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))

plt.bar(df['state'], df['yield_kg_per_ha'], color='violet')

plt.title('Crop yield by state (practice data)')
plt.xlabel('State')
plt.ylabel('Yield (kg/ha)')
plt.tight_layout()
plt.show()
```

```python
# Load the crop production dataset
crop_df = pd.read_csv('Crop_Production_Data.csv')

# First look - always do these 5 things with any new dataset
print("Shape:", crop_df.shape)
print("\nColumn names:")
print(crop_df.columns.tolist())
print("\nFirst 5 rows:")
crop_df.head()
```

```python
print("=== DATA HEALTH CHECK===")
print("\n1. Shape (rows, columns):", crop_df.shape)
print("\n2. column data types:")
print(crop_df.dtypes)
print("\n3. Missing values per column:")
print(crop_df.isnull().sum())
print("\n4. Basic statistics:")
crop_df.describe()
```

```python
rain_df = pd.read_csv('rainfall_data_1901to2017.csv')

print("Shape:", rain_df.shape)
print("\nColumns:", rain_df.columns.tolist())
print("\nFirst 5 rows:")
rain_df.head()
```

```python
# health check

print("Missing values in rainfall data:")
print(rain_df.isnull().sum())
print("\nYear rang:", rain_df['YEAR'].min(), "to", rain_df['YEAR'].max())
print("Unique states/region:", rain_df['SUBDIVISION'].unique())
```

```python
temp_df = pd.read_csv('temperature.csv')

print("Shape:", temp_df.shape)
print("\nColumns:", temp_df.columns.tolist())
print("\nMissing values in temperature data:")
print(temp_df.isnull().sum())
```

```python
temp_df.head()
```

```python
water_df = pd.read_csv('Indian_water_data.csv')

print("Shape", water_df.shape)
print("\nColumns:", water_df.columns.tolist())
print("\nMissing values in water data:")
print(water_df.isnull().sum())
water_df.head()
```

```python
if 'Yield' not in crop_df.columns:
    crop_df['Yield'] = crop_df['Production'] / crop_df['Area']
    print("Yield column created!")
else:
    print("Yield column already exists")
```

***# Average yield of Rice vs Wheat across all states and years***

```python
avg_yield = crop_df[crop_df['Crop'].isin(['Rice', 'Wheat'])]\
            .groupby('Crop')['Yield'].mean()

print(avg_yield)

plt.figure(figsize=(6, 4))
avg_yield.plot(kind='bar', color=['#2196F3', '#FF9800'], edgecolor='white')
plt.title('Average yield: Rice vs Wheat (all states, all years)')
plt.ylabel('Yield (kg/ha)')
plt.xlabel('crop')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
